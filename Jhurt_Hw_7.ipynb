Exercise 16.1 - Fashion MNIST

```python
from tensorflow.keras.datasets import fashion_mnist
(x_train, y_train), (x_test, y_test) = fashion_mnist.load_data()
```

```python
print(x_train.shape)
print(y_train.shape)
print(x_test.shape)
print(y_test.shape)
```

```python
x_train = x_train.reshape((60000, 28, 28, 1))
x_test = x_test.reshape((10000, 28, 28, 1))

print(x_train.shape)
print(x_test.shape)
```

```python
x_train=x_train.astype('float32')/255
x_test=x_test.astype('float32')/255
```

```python
from tensorflow.keras.utils import to_categorical
y_train = to_categorical(y_train)
y_test = to_categorical(y_test)

print(y_train.shape)
print(y_test.shape)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, Dense, Flatten, MaxPooling2D

cnn=Sequential()

cnn.add(Conv2D(filters=64, kernel_size=(3,3), activation='relu', input_shape=(28,28,1)))
cnn.add(MaxPooling2D(pool_size=(2,2)))
cnn.add(Conv2D(filters=128, kernel_size=(3,3), activation='relu'))
cnn.add(MaxPooling2D(pool_size=(2,2)))
cnn.add(Flatten())
cnn.add(Dense(units=128, activation='relu'))
cnn.add(Dense(units=10, activation='softmax'))

```

```python
cnn.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
```

```python
cnn.summary()
```

```python
import time
start_time = time.perf_counter()
cnn.fit(x_train, y_train, epochs=5, batch_size=64, validation_split=0.1)

fashion_time=time.perf_counter()-start_time
print('Training Time:', fashion_time, 'seconds')
```

```python
loss, fashion_accuracy = cnn.evaluate(x_test, y_test)
print('Test Accuracy:', fashion_accuracy)
```

MNIST Comparison

```python
from tensorflow.keras.datasets import mnist
(x_train, y_train), (x_test, y_test) = mnist.load_data()

print (x_train.shape)
print (y_train.shape)
print (x_test.shape)
print (y_test.shape)
```

```python
x_train = x_train.reshape((60000, 28, 28, 1))
x_test = x_test.reshape((10000, 28, 28, 1))

x_train=x_train.astype('float32')/255
x_test=x_test.astype('float32')/255

y_train = to_categorical(y_train)
y_test = to_categorical(y_test)
```

```python
mnist_cnn=Sequential()
mnist_cnn.add(Conv2D(filters=64, kernel_size=(3,3), activation='relu', input_shape=(28,28,1)))
mnist_cnn.add(MaxPooling2D(pool_size=(2,2)))

mnist_cnn.add(Conv2D(filters=128, kernel_size=(3,3), activation='relu'))
mnist_cnn.add(MaxPooling2D(pool_size=(2,2)))

mnist_cnn.add(Flatten())
mnist_cnn.add(Dense(units=128, activation='relu'))
mnist_cnn.add(Dense(units=10, activation='softmax'))

mnist_cnn.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
```

```python
start_time = time.perf_counter()
mnist_cnn.fit(x_train, y_train, epochs=5, batch_size=64, validation_split=0.1)

mnist_time=time.perf_counter()-start_time

print('Training Time:', mnist_time, 'seconds')
```

```python
loss, mnist_accuracy = mnist_cnn.evaluate(x_test, y_test)
print('Test Accuracy:', mnist_accuracy)
```

**Exercise 16.1 Results:** The model performed better on MNIST than Fashion-MNIST. The test accuracy for MNIST was 98.69%, while the test accuracy for Fashion-MNIST was 90.04%. The training times were very close. Fashion-MNIST took 26.51 seconds to train, while MNIST took 26.83 seconds.

4096-Neuron Dense Layer

```python
large_dense_cnn=Sequential()
large_dense_cnn.add(Conv2D(filters=64, kernel_size=(3,3), activation='relu', input_shape=(28,28,1)))
large_dense_cnn.add(MaxPooling2D(pool_size=(2,2)))
large_dense_cnn.add(Conv2D(filters=128, kernel_size=(3,3), activation='relu'))
large_dense_cnn.add(MaxPooling2D(pool_size=(2,2)))
large_dense_cnn.add(Flatten())
large_dense_cnn.add(Dense(units=4096, activation='relu'))
large_dense_cnn.add(Dense(units=128, activation='relu'))
large_dense_cnn.add(Dense(units=10, activation='softmax'))

large_dense_cnn.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
```

```python
large_dense_cnn.summary()
```

```python
start_time=time.perf_counter()
large_dense_cnn.fit(x_train, y_train, epochs=5, batch_size=64, validation_split=0.1)
large_dense_time=time.perf_counter()-start_time

print('Training Time:', large_dense_time, 'seconds')
```

```python
loss, large_dense_accuracy = large_dense_cnn.evaluate(x_test, y_test)
print('Test Accuracy:', large_dense_accuracy)
```

**Exercise 16.4 Results:** Removing the 1st Dense layer slightly decreased the test accuracy from 98.69% to 98.58%. It also decreased the training time from 26.83 seconds to 25.41 seconds. Adding a Dense layer with 4096-neurons increased the test accuracy to 99.10%, but it also increased the training time to 42.93 seconds. The 4096-neuron model had the highest accuracy, but it was also the slowest model to train.
